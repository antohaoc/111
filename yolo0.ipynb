```python
!pip -q install -U ultralytics
from ultralytics import YOLO

model = YOLO("yolo11n-seg.pt")
model.train(
    data="carparts-seg.yaml",
    epochs=100,
    patience=20,
    batch=16,
    imgsz=640,
    lr0=0.01,
    optimizer="AdamW",
    freeze=0,
    seed=0,
    project="runs/freeze-carparts",
    name="yolo11n_freeze0",
)
```
