```python
import glob
import torch
import pandas as pd
from ultralytics import YOLO

paths = sorted(glob.glob("/content/drive/MyDrive/magisterka/**/best.pt", recursive=True))
print(f"знайдено ваг: {len(paths)}")
for p in paths:
    print(p)

rows = []
for weight_path in paths:
    name = weight_path.split("/")[-3]
    print(f"-> {name}")
    model = YOLO(weight_path)
    res = model.val(data="carparts-seg.yaml", split="val", imgsz=640, batch=16, verbose=False)
    names = res.names
    seg = res.seg
    for i, class_idx in enumerate(seg.ap_class_index):
        rows.append({
            "Клас": names[int(class_idx)],
            "Модель": name,
            "Precision": round(float(seg.p[i]), 3),
            "Recall": round(float(seg.r[i]), 3),
            "mAP50": round(float(seg.ap50[i]), 3),
            "mAP50-95": round(float(seg.ap[i]), 3),
        })
    torch.cuda.empty_cache()

df = pd.DataFrame(rows)
out = "/content/drive/MyDrive/magisterka/Per_Class_Metrics.xlsx"
df.to_excel(out, index=False)
print(out)
print(df.groupby("Модель")[["mAP50", "Recall"]].mean())
```

```python
import os
import glob
import torch
import pandas as pd
from ultralytics import YOLO

print("Починаємо збір покласових метрик для таблиці...")

# 1. Функція пошуку ваг
def find_weights(model_name):
    paths = glob.glob(f"/content/runs/**/{model_name}/weights/best.pt", recursive=True)
    return paths[0] if paths else None

model_names = [
    "yolov8n_freeze0",
    "yolo11n_freeze0", "yolo11n_freeze6", "yolo11n_freeze10",
    "yolo26n_freeze0", "yolo26n_freeze6", "yolo26n_freeze10"
]

all_metrics = []

# 2. Збір даних
for m_name in model_names:
    weight_path = find_weights(m_name)
    if not weight_path:
        print(f"[!] Ваги для {m_name} не знайдено. Пропускаємо.")
        continue

    print(f"-> Обробка моделі: {m_name}...")
    try:
        model = YOLO(weight_path)
        res = model.val(data="carparts-seg.yaml", split="val", imgsz=640, batch=16, verbose=False)
        names = res.names
        seg = res.seg

        for i, class_idx in enumerate(seg.ap_class_index):
            cls_name = names[class_idx]
            all_metrics.append({
                "Деталь (Клас)": cls_name,
                "Модель": m_name,
                "Recall (Повнота)": round(seg.r[i], 3),
                "mAP@50 (Точність)": round(seg.map50[i], 3)
            })
        torch.cuda.empty_cache()
    except Exception as e:
        print(f"Помилка при обробці {m_name}: {e}")

# 3. Формування та збереження файлу
if all_metrics:
    df = pd.DataFrame(all_metrics)
    pivot_df = df.pivot(index="Деталь (Клас)", columns="Модель", values=["Recall (Повнота)", "mAP@50 (Точність)"])

    excel_path = "/content/Per_Class_Metrics_Diploma.xlsx"
    pivot_df.to_excel(excel_path)
    print(f"\n[OK] Excel-файл успішно згенеровано: {excel_path}")

    # 4. Автоматичне копіювання на Google Диск
    drive_dir = "/content/drive/MyDrive/magisterka/"
    os.makedirs(drive_dir, exist_ok=True)
    os.system(f"cp {excel_path} {drive_dir}")
    print(f"[OK] Файл Per_Class_Metrics_Diploma.xlsx гарантовано збережено на Google Диск у папку magisterka!")
else:
    print("\n[Помилка] Не вдалося зібрати жодних метрик. Перевірте, чи є навчені моделі в Colab.")
```

```python
import matplotlib.pyplot as plt
import cv2
import random

print("=== ЕТАП 2: ГЕНЕРАЦІЯ ПОРІВНЯЛЬНИХ КОЛАЖІВ ===")

# Створюємо папку для збереження готових картинок
out_dir = "/content/visual_comparisons"
os.makedirs(out_dir, exist_ok=True)

# Обираємо 6 моделей для порівняння (як просила керівниця)
compare_list = [
    "yolo11n_freeze0", "yolo11n_freeze6", "yolo11n_freeze10",
    "yolo26n_freeze0", "yolo26n_freeze6", "yolo26n_freeze10"
]

# Завантажуємо моделі в пам'ять
loaded_models = {}
for m_name in compare_list:
    w_path = find_weights(m_name)
    if w_path:
        loaded_models[m_name] = YOLO(w_path)

# Знаходимо тестові зображення
test_images = glob.glob("/content/datasets/carparts-seg/images/test/*.*")
random.seed(42) # Щоб картинки щоразу обиралися однакові
sample_images = random.sample(test_images, min(15, len(test_images)))

for idx, img_path in enumerate(sample_images, 1):
    img_name = os.path.basename(img_path)
    print(f"Обробка фото {idx}/15: {img_name}...")

    # Створюємо сітку 2 рядки, 3 стовпці
    fig, axes = plt.subplots(2, 3, figsize=(24, 14))
    axes = axes.flatten()

    for ax, m_name in zip(axes, compare_list):
        if m_name not in loaded_models:
            continue

        model = loaded_models[m_name]
        # Прогнозуємо (conf=0.25 - стандартний поріг впевненості)
        results = model.predict(img_path, imgsz=640, conf=0.25, verbose=False)[0]

        # Малюємо маски поверх фото
        annotated_img = results.plot(line_width=2, font_size=12)
        annotated_rgb = cv2.cvtColor(annotated_img, cv2.COLOR_BGR2RGB)

        ax.imshow(annotated_rgb)
        ax.set_title(m_name.replace("_", " "), fontsize=18, fontweight='bold')
        ax.axis("off")

    plt.tight_layout()
    save_path = os.path.join(out_dir, f"Comparison_{idx}_{img_name}")
    plt.savefig(save_path, dpi=200, bbox_inches='tight')
    plt.close()

print(f"\n[OK] 15 колажів успішно згенеровано у папці: {out_dir}")

# === КОПІЮВАННЯ НА GOOGLE ДИСК ===
print("\nКопіювання результатів на Google Диск...")
!cp /content/Per_Class_Metrics_Diploma.xlsx /content/drive/MyDrive/magisterka/
!cp -r /content/visual_comparisons /content/drive/MyDrive/magisterka/
print("[OK] Усі дані збережено на Вашому Google Диску в папці magisterka!")
```
